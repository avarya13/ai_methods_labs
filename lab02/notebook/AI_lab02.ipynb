```python
import numpy as np
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
def load_tokenizer_and_model(model_name_or_path):
  return GPT2Tokenizer.from_pretrained(model_name_or_path), GPT2LMHeadModel.from_pretrained(model_name_or_path) #.cuda()
```

```python
def generate(
    model, tok, text, num_return_sequences=3,
    do_sample=True, max_length=250, repetition_penalty=5.0,
    top_k=5, top_p=0.95, temperature=1,
    num_beams=5,
    no_repeat_ngram_size=3
):
    input_ids = tok.encode(text, return_tensors="pt")#.cuda()
    outputs = []

    for i in range(num_return_sequences):
        out = model.generate(
            input_ids, #.cuda()
            max_length=max_length,
            repetition_penalty=repetition_penalty,
            no_repeat_ngram_size=no_repeat_ngram_size,
            do_sample=do_sample,
            top_k=top_k + 5*i,
            top_p=top_p,
            temperature=temperature + 0.1 * i,
        )
        outputs.append(tok.decode(out[0], skip_special_tokens=True))

    return outputs
```

```python
tok, model = load_tokenizer_and_model("sberbank-ai/rugpt3large_based_on_gpt2")
```

```python
text_annotation = 'Актуальность сегментация клубочков обонятельной луковицы мозга человека заключается в'
```

```python
text_intro = 'Сегментация клубочков обонятельной луковицы человека на рентгеновских фазо-контрастных томографических снимках представляет собой важную задачу в'
```

```python
text_method = 'Для сегментации клубочков обонятельной луковицы был разработан метод, основанный на архитектуре U-Net, который'
```

```python
text_conclusion = 'Исследование подтверждает эффективность применения U-Net для сегментации клубочков обонятельной луковицы, что'
```

```python
generate(model, tok, text_annotation)
```

```python
generate(model, tok, text_intro)
```

```python
generate(model, tok, text_method)
```

```python
generate(model, tok, text_conclusion)
```
